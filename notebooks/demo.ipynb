### 自定义数据:

将您的阿拉伯语地址数据按照以下格式准备:

```json
[
  {
    "text": "阿拉伯语地址文本",
    "entities": [
      {
        "start": 0,
        "end": 10,
        "label": "STREET",
        "text": "实体文本"
      }
    ]
  }
]
```

支持的实体类型:
- STREET: 街道
- BUILDING: 建筑物
- DISTRICT: 地区
- CITY: 城市
- COUNTRY: 国家
- POSTAL_CODE: 邮政编码
