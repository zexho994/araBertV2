# NER 训练,评估启动器

```python
from typing import Optional
import sys
import os
from pathlib import Path


def find_project_root(start: Path) -> Path:
    for path in [start, *start.parents]:
        if (path / 'ner_cli.py').exists() and (path / 'src').exists():
            return path
    return start


project_root = find_project_root(Path.cwd())
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

print(f'当前工作目录: {os.getcwd()}')
print(f'项目根目录: {project_root}')


def exec_train_command(country: str) -> int:
    """调试训练命令（Notebook 版本）"""
    import sys
    import os

    # 模拟命令行参数
    sys.argv = [
        'ner_cli.py',
        '--verbose',
        'train',
        '--country',
        country,
    ]

    print(f"模拟命令行参数: {' '.join(sys.argv)}")
    print(f"当前工作目录: {os.getcwd()}")
    print(f"项目根目录: {project_root}")

    # 切换到项目根目录，确保相对路径与 CLI 一致
    os.chdir(project_root)

    try:
        import ner_cli
        print("开始执行训练命令...")
        print("=" * 50)
        result = ner_cli.main()
        print("=" * 50)
        print(f"命令执行完成，返回码: {result}")
        return result
    except Exception as e:
        print(f"执行过程中出现错误: {e}")
        import traceback
        traceback.print_exc()
        return 1


def exec_evaluate_command(country: str) -> int:
    """调试评估命令（Notebook 版本）"""
    import sys
    import os

    model_path = f'data/ner/models/{country}/best_model'
    data_path = f'data/ner/data/{country}/val.jsonl'

    sys.argv = [
        'ner_cli.py',
        '--verbose',
        'evaluate',
        '--country',
        country,
        '--model-path',
        model_path,
        '--data-path',
        data_path,
    ]

    print(f"模拟命令行参数: {' '.join(sys.argv)}")
    print(f"当前工作目录: {os.getcwd()}")
    print(f"项目根目录: {project_root}")

    os.chdir(project_root)

    try:
        import ner_cli
        print("开始执行评估命令...")
        print("=" * 50)
        result = ner_cli.main()
        print("=" * 50)
        print(f"命令执行完成，返回码: {result}")
        return result
    except Exception as e:
        print(f"执行过程中出现错误: {e}")
        import traceback
        traceback.print_exc()
        return 1


country = 'uae_xml_roberta_base'
model = 'train' # train or eval
# model = 'eval' # train or eval

if model == 'train':
    exit_code = exec_train_command(country)
elif model == 'eval':
    exit_code = exec_evaluate_command(country)
else:
    raise ValueError("model 必须是 'train' 或 'eval'")

exit_code
```
